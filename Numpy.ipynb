# Array Stacking and splitting

```python
#Stack
a = np.array([1,2,3])
b = np.array([4,5,6])

```

```python
print(np.vstack((a,b)))
print(np.hstack((a,b)))
```

```python
#Split
c = np.array([[1,2,3],[4,5,6]])
print(c)
```

```python
hsplit = np.hsplit(c,3)

for s in hsplit:
    print(s)
```

```python
vsplit = np.vsplit(c,2)

for s in vsplit:
    print(s)
```

# Matheamtical operation on Multiple Array

```python
a= np.array([1,2,3])
b= np.array([4,5,6])
print(a)
print(b)
```

```python
print(np.add(a,b))
print(np.subtract(a,b))
print(np.multiply(a,b))
print(np.divide(a,b))
```

```python
print(np.dot(a,b))
```

```python
#Transpose
t = np.array([[1,2,3]
             ,[4,5,6]])
print(t)
print(t.T)
```

```python
# Statistical Function
a= np.array([[1,2,3],[4,5,6]])
print(a)
```

```python
print(np.sum(a))
print(np.mean(a))
print(np.median(a))
print(np.std(a))
print(np.max(a))
print(np.min(a))
```
